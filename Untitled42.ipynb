```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import shap


class BaseModel:
    def __init__(self, name, random_state=42):
        self.name = name
        self.random_state = random_state
        self.model = None
        self.target_name = None

    def _build_model(self):
        raise NotImplementedError

    def fit(self, X, y):
        if self.model is None:
            self._build_model()
        self.model.fit(X, y)
        return self

    def predict(self, X):
        if self.model is None:
            raise RuntimeError("Model not trained yet.")
        return self.model.predict(X)

    def feature_importance(self, feature_names, save_path):
        os.makedirs(os.path.dirname(save_path) or ".", exist_ok=True)
        importances = self.model.feature_importances_
        idx = np.argsort(importances)
        fig, ax = plt.subplots(figsize=(8, max(4, len(feature_names) * 0.4)))
        ax.barh(range(len(feature_names)), importances[idx])
        ax.set_yticks(range(len(feature_names)))
        ax.set_yticklabels([feature_names[i] for i in idx])
        ax.set_xlabel("Importance")
        ax.set_title(f"{self.name} - {self.target_name} Feature Importance")
        plt.tight_layout()
        fig.savefig(save_path, dpi=150, bbox_inches="tight")
        plt.close(fig)

    def shap_analysis(self, X_test, feature_names, save_path):
        os.makedirs(os.path.dirname(save_path) or ".", exist_ok=True)
        try:
            X_df = pd.DataFrame(X_test, columns=feature_names)
            explainer = shap.TreeExplainer(self.model)
            shap_values = explainer.shap_values(X_df, check_additivity=False)
            fig, ax = plt.subplots(figsize=(10, max(4, len(feature_names) * 0.4)))
            shap.summary_plot(shap_values, X_df, show=False)
            plt.tight_layout()
            fig.savefig(save_path, dpi=150, bbox_inches="tight")
            plt.close(fig)
            n_samples = min(100, X_df.shape[0])
            fig2, ax2 = plt.subplots(figsize=(8, max(4, len(feature_names) * 0.4)))
            shap.summary_plot(shap_values[:n_samples], X_df.iloc[:n_samples], plot_type="bar", show=False)
            plt.tight_layout()
            bar_path = save_path.replace(".png", "_bar.png")
            fig2.savefig(bar_path, dpi=150, bbox_inches="tight")
            plt.close(fig2)
        except Exception as e:
            fig, ax = plt.subplots(figsize=(6, 4))
            ax.text(0.5, 0.5, f"SHAP failed: {e}", ha="center", va="center", transform=ax.transAxes)
            fig.savefig(save_path, dpi=150, bbox_inches="tight")
            plt.close(fig)
```

```python
import xgboost as xgb
from sklearn.model_selection import train_test_split, KFold, RandomizedSearchCV
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

if __name__ == "__main__":
    # For Jupyter/Colab, use the current working directory
    BASE_DIR = os.getcwd()

    CONFIG = {
        "dataset_path": os.path.join(
            BASE_DIR, "CsSnI3_dataset.csv"
        ),
        "test_size": 0.2,
        "random_state": 42,
        "cv_folds": 5,
        "search_iterations": 30,
        "output_dir": os.path.join(
            BASE_DIR, "output", "xgboost"
        ),
        "output_file": "results_xgboost.csv"
    }

    OUTPUT_DIR = CONFIG["output_dir"]
    os.makedirs(OUTPUT_DIR, exist_ok=True)

    # The xgb import was missing earlier, moved it to the top. Removed this check.

    # Load dataset
    df = pd.read_csv(CONFIG["dataset_path"], encoding='latin1')

    # Rename columns for easier access and consistency with feature/target lists
    df = df.rename(columns={
        'Thickness (µm)': 'Thickness',
        'Defect Density (cm^-3)': 'DefectDensity',
        'Acceptor Density (cm^-3)': 'DopingDensity',
        'Voc (V)': 'Voc',
        'Jsc (mA/cm^2)': 'Jsc',
        'PCE (%)': 'PCE'
    })

    # Log transformations
    if "DopingDensity" in df.columns:
        df["log_DopingDensity"] = np.log10(df["DopingDensity"])

    if "DefectDensity" in df.columns:
        df["log_DefectDensity"] = np.log10(df["DefectDensity"])

    # One-hot encode Material
    if "Material" in df.columns:
        df = pd.get_dummies(
            df,
            columns=["Material"],
            prefix=["Mat"]
        )

    # Material feature columns
    mat_cols = [
        col for col in df.columns
        if col.startswith("Mat_")
    ]

    # Feature columns
    feature_cols = [
        "Thickness",
        "log_DopingDensity",
        "log_DefectDensity"
    ] + mat_cols

    # Keep only columns that actually exist
    feature_cols = [
        c for c in feature_cols
        if c in df.columns
    ]

    # Target columns
    targets = [
        t for t in ["Voc", "Jsc", "FF", "PCE"]
        if t in df.columns
    ]

    print("\n" + "=" * 60)
    print("ZERO-VARIANCE FEATURE REPORT")
    print("=" * 60)

    cols_to_remove = []

    for col in feature_cols:
        num_unique = df[col].nunique()

        if num_unique == 1:
            print(
                f"Column: {col}, "
                f"Unique Values: {num_unique} "
                f"-> REMOVE (zero variance)"
            )
            cols_to_remove.append(col)
        else:
            print(
                f"Column: {col}, "
                f"Unique Values: {num_unique} "
                f"-> KEEP"
            )

    # Remove zero-variance features
    feature_cols = [
        col for col in feature_cols
        if col not in cols_to_remove
    ]

    print("\nFinal Feature List after removing zero-variance columns:")
    print(feature_cols)

    # Hyperparameter search space
    param_dist = {
        "n_estimators": [100, 200, 300, 400, 500],
        "learning_rate": [0.01, 0.03, 0.05, 0.1, 0.2],
        "max_depth": [3, 5, 7, 10, 15],
        "min_child_weight": [1, 3, 5, 10],
        "subsample": [0.6, 0.8, 1.0],
        "colsample_bytree": [0.6, 0.8, 1.0],
        "gamma": [0, 0.1, 0.2, 0.5]
    }

    results = []

    # Train one model for each target
    for target in targets:

        print("\n" + "=" * 60)
        print("TARGET:", target)
        print("=" * 60)

        X = df[feature_cols]
        y = df[target]

        # Train/test split
        X_train, X_test, y_train, y_test = train_test_split(
            X,
            y,
            test_size=CONFIG["test_size"],
            random_state=CONFIG["random_state"]
        )

        # Base XGBoost estimator
        estimator = xgb.XGBRegressor(
            random_state=CONFIG["random_state"],
            verbosity=0
        )

        # Randomized hyperparameter search
        search = RandomizedSearchCV(
            estimator,
            param_distributions=param_dist,
            n_iter=CONFIG["search_iterations"],
            cv=KFold(
                CONFIG["cv_folds"],
                shuffle=True,
                random_state=CONFIG["random_state"]
            ),
            scoring="r2",
            random_state=CONFIG["random_state"],
            n_jobs=-1
        )

        print("\nRunning hyperparameter search...")

        search.fit(X_train, y_train)

        best_xgb = search.best_estimator_

        # Predictions
        pred_train = best_xgb.predict(X_train)
        pred_test = best_xgb.predict(X_test)

        # Metrics
        train_r2 = r2_score(y_train, pred_train)
        test_r2 = r2_score(y_test, pred_test)

        train_rmse = np.sqrt(
            mean_squared_error(y_train, pred_train)
        )
        test_rmse = np.sqrt(
            mean_squared_error(y_test, pred_test)
        )

        train_mae = mean_absolute_error(
            y_train,
            pred_train
        )
        test_mae = mean_absolute_error(
            y_test,
            pred_test
        )

        # Display parameters
        print("\nBest Parameters")
        print(search.best_params_)

        # Display metrics
        print("\nMetrics")
        print("Train R2  :", train_r2)
        print("Test R2   :", test_r2)
        print("Train RMSE:", train_rmse)
        print("Test RMSE :", test_rmse)
        print("Train MAE :", train_mae)
        print("Test MAE  :", test_mae)

        # Store results
        results.append([
            target,
            train_r2,
            test_r2,
            train_rmse,
            test_rmse,
            train_mae,
            test_mae,
            str(search.best_params_)
        ])

        # ==========================================================
        # FEATURE IMPORTANCE
        # ==========================================================

        imp_path = os.path.join(
            OUTPUT_DIR,
            f"{target}_xgb_feature_importance.png"
        )

        plt.figure(figsize=(8, 5))

        importance = pd.Series(
            best_xgb.feature_importances_,
            index=feature_cols
        )

        importance.sort_values().plot.barh()

        plt.title(f"{target} Feature Importance")
        plt.xlabel("Importance")
        plt.tight_layout()

        plt.savefig(
            imp_path,
            dpi=300,
            bbox_inches="tight"
        )

        plt.close()

        print(f"Saved: {imp_path}")

        # ==========================================================
        # PARITY PLOT
        # ==========================================================

        parity_path = os.path.join(
            OUTPUT_DIR,
            f"{target}_xgb_parity_plot.png"
        )

        plt.figure(figsize=(5, 5))

        plt.scatter(
            y_test,
            pred_test,
            alpha=0.7
        )

        mn = min(
            y_test.min(),
            pred_test.min()
        )

        mx = max(
            y_test.max(),
            pred_test.max()
        )

        plt.plot(
            [mn, mx],
            [mn, mx],
            "r--"
        )

        plt.xlabel("Actual")
        plt.ylabel("Predicted")
        plt.title(f"{target} Parity Plot")

        plt.tight_layout()

        plt.savefig(
            parity_path,
            dpi=300,
            bbox_inches="tight"
        )

        plt.close()

        print(f"Saved: {parity_path}")

        # ==========================================================
        # SHAP ANALYSIS
        # ==========================================================

        shap_path = os.path.join(
            OUTPUT_DIR,
            f"{target}_xgb_shap_summary.png"
        )

        try:
            explainer = shap.TreeExplainer(best_xgb)

            shap_values = explainer.shap_values(
                X_test,
                check_additivity=False
            )

            shap.summary_plot(
                shap_values,
                X_test,
                feature_names=feature_cols,
                show=False
            )

            plt.savefig(
                shap_path,
                dpi=300,
                bbox_inches="tight"
            )

            plt.close()

            print(f"Saved: {shap_path}")

        except Exception as e:
            print(f"SHAP analysis failed for {target}: {e}")

    # ==============================================================
    # FINAL RESULTS
    # ==============================================================

    results_df = pd.DataFrame(
        results,
        columns=[
            "Target",
            "Train_R2",
            "Test_R2",
            "Train_RMSE",
            "Test_RMSE",
            "Train_MAE",
            "Test_MAE",
            "Best_Parameters"
        ]
    )

    print("\n" + "=" * 60)
    print("FINAL RESULTS")
    print("=" * 60)

    print(results_df.to_string(index=False))

    # Save results
    results_path = os.path.join(
        OUTPUT_DIR,
        CONFIG["output_file"]
    )

    results_df.to_csv(
        results_path,
        index=False
    )

    print(f"\nResults saved to: {results_path}")
```
